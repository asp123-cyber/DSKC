

# 1.   **UNIFORM DISTRIBUTION**




```python
#Importing Required Libraries
import numpy as np
import matplotlib.pyplot as plt

#To Generate Random Numbers from Unifrom Distribution
n1=1000
x1=np.random.uniform(0,1,n1)
n2=5000
x2=np.random.uniform(0,1,n2)
n3=10000
x3=np.random.uniform(0,1,n3)

#Plotting for different bin sizes
plt.figure(figsize=(12, 6))
plt.subplot(1,3,1)
plt.hist(x1,bins=40, edgecolor='black', color='red', label='40 Bins')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.subplot(1,3,2)
plt.hist(x1,bins=20, edgecolor='black', color='blue', label='20 Bins')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.title('Uniform Distribution with different Bin sizes')
plt.legend()
plt.subplot(1,3,3)
plt.hist(x1,bins=30, edgecolor='black', color='orange', label='30 Bins')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.show()

#Plotting for different Number of points
plt.figure(figsize=(12, 6))
plt.subplot(1,3,1)
plt.hist(x1,bins=30, edgecolor='black', color='red', label='1000 Points')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.subplot(1,3,2)
plt.hist(x2,bins=30, edgecolor='black', color='blue', label='5000 Points')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.title('Uniform Distribution with different Number of points')
plt.legend()
plt.subplot(1,3,3)
plt.hist(x3,bins=30, edgecolor='black', color='orange', label='10000 Points')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

# Central Limit Theorem Verification



*   Poisson Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.random.poisson(10, 1000000)       #Random Numbers from Poisson Distribution
y=np.array_split(x, 1000)              #Spliting into 1000 different arrays

meani=[]
stdi=[]
for i in range(len(y)):
  meani.append(np.mean(y[i]))          #Making a list of means of all 1000 arrays

meanm=np.mean(meani)                   #Mean of all means
stdm=np.std(meani)                     #Standard deviation of all means
meanp=np.mean(x)                       #Mean of whole distribution
stdp=np.std(x)                         #Standard deviation of whole distribution
print(meanm, meanp)
print(stdm,stdp/(np.sqrt(1000)))

plt.figure(figsize=(12, 6))
plt.subplot(1,2,1)
plt.hist(x, 30)
plt.xlabel('Poisson Distributions')
plt.ylabel('Frequency')
plt.subplot(1,2,2)
plt.hist(meani, 40)
plt.xlabel('Mean of Poisson Distributions')
plt.ylabel('Frequency')
plt.title("Gaussian Distribution of Means")
plt.show()
```
