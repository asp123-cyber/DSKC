#1.   **VALUE OF PI**






*   **Using a Circle inside a Square**



```python
#Importing Required Libraries
import numpy as np
import matplotlib.pyplot as plt

n=10000 #Number of points

#Defining required lists
X1=[]
X2=[]
Y1=[]
Y2=[]
PI=[]

nc=0
for i in range(1,n+1):
  x=np.random.uniform(0,1) #GENERATING RANDOM VALUE OF X AND Y
  y=np.random.uniform(0,1)
  if x**2+y**2<=1:         #Condition for points lying inside the Circle
    nc+=1                  #Points Inside the Circle
    X1.append(x)
    Y1.append(y)
  else:
    X2.append(x)
    Y2.append(y)
  if i % 100 == 0: #To only Compute for every 100th point
      pi=4*(nc/i)  #Value of PI after ith point
      PI.append(pi)

N = list(range(100, n + 1, 100))

#For plotting the points and Monte Carlo Simulation in same fig
plt.figure(figsize=(12, 6))
plt.subplot(1,2,1)
plt.scatter(X1,Y1,s=1,color='blue', label='Points Inside the Circle')
plt.scatter(X2,Y2,s=1,color='red', label='Points Outside the Circle')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Points inside the square")
plt.legend()
plt.tight_layout()
plt.grid()
plt.subplot(1,2,2)
plt.plot(N,PI,color='red')
plt.xlabel("Number of Points")
plt.ylabel("Value of PI using MC simulation")
plt.title("Monte Carlo Simulation for a Circle")
plt.axhline(y=np.pi, ls='dashed',label='Actual value of PI')
plt.tight_layout()
plt.legend()
plt.grid()
plt.show()

print("Value of Pi after 10000 points: ",PI[-1])
```



*  **Using a Sphere inside a Box**



```python
#Importing Required Libraries
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

n=10000

PI=[]
X1=[]
X2=[]
Y1=[]
Y2=[]
Z1=[]
Z2=[]
nc=0
for i in range(1,n+1):
  x=np.random.uniform(0,1)
  y=np.random.uniform(0,1)
  z=np.random.uniform(0,1)
  if x**2+y**2+z**2<=1: #Condition for Points lying inside the Sphere
    nc+=1
    X1.append(x)
    Y1.append(y)
    Z1.append(z)
  else:
    X2.append(x)
    Y2.append(y)
    Z2.append(z)               #Points Inside the Sphere
  if i % 100 == 0:
      pi=6*(nc/i)       #Value of Pi after ith point
      PI.append(pi)

N = list(range(100, n + 1, 100))

#3d Plot of Random Numbers
fig = go.Figure(data=[go.Scatter3d(x=X1, y=Y1, z=Z1,
                                   mode='markers', name='Points inside the Sphere', marker=dict(color='Blue', size=2)), go.Scatter3d(x=X2, y=Y2, z=Z2,
                                   mode='markers', name='Points outside the Sphere', marker=dict(color='Red', size=2))])
fig.update_layout(
        title='Points Inside the Cube',
        scene = dict(
            xaxis = dict(title='X'),
            yaxis = dict(title='Y'),
            zaxis = dict(title='Z')
        )
    )
fig.show()
#Plotting MC Simulation for the Sphere
plt.plot(N,PI,color='red')
plt.xlabel("Number of Points")
plt.ylabel("Value of PI using MC simulation")
plt.title("Monte Carlo Simulation for a Sphere")
plt.axhline(y=np.pi, ls='dashed',label='Actual value of PI')
plt.tight_layout()
plt.legend()
plt.grid()
plt.show()

print("Value of Pi after 10000 points: ",PI[-1])
```

Pudding Cost


```python
import numpy as np
PRICE=[]
nt=10000  #Total Number of Puddings to be made

for n in range(1,nt + 1):
  MILK=np.random.uniform(18,20)                         #Random value from Uniform distribution
  EGG=np.random.choice([2,2.5,3], 1, p=[0.2,0.5,0.3])   #Random value from Discrete distribution
  SUGAR=np.random.uniform(20,23)
  BREAD=np.random.normal(25,1)                          #Random value from Normal distribution
  ALMONDS=np.random.choice([700,500], 1, p=[0.7,0.3])
  PRICE.append(MILK + 2*EGG + SUGAR/10 + BREAD/6 + ALMONDS/10 + 5)

AVG_PRICE=sum(PRICE)/len(PRICE) #Average value for 10000 Puddings
print(f"Average price of the Pudding for Profit of 5rs is: {AVG_PRICE[0]}rs")
```
