

# 2.   **MONTE CARLO SINGLE INTEGRAL**





*   **HIT OR MISS**



```python
#Importing Required Libraries
from scipy.integrate import quad
import numpy as np
import matplotlib.pyplot as plt

#Limits of Integration by user Input
a=int(input("Enter lower limit a: "))
b=int(input("Enter Upper limit b: "))

#Defining function whose Integral is to be found
def f(x):
  return 1/(1+x**2)

nc=0
nt=10000 #Total Number of Points

#To find the maximum value of y
xi=np.linspace(a,b,nt)
yi=f(xi)
M=np.max(yi)

X1=[]
X2=[]
Y1=[]
Y2=[]

for n in range(1,nt + 1):
  x=np.random.uniform(a,b)
  y=np.random.uniform(0,M)
  if y<=f(x):   #Required Condition for Monte Carlo Simulation
    X1.append(x)
    Y1.append(y)
    nc+=1       #Points Under the Curve
  else:
    X2.append(x)
    Y2.append(y)

func=(b-a)*M*(nc/nt)
print("Integral Solution from monte carlo: ", func)
c,_=quad(f, a, b)           #Solving analytically using quad function
Error=(abs(func-c)/c)*100   #To calculate the associated Percentage Error
print("Analytical Solution: ", c)
print(f"Error: {Error}%")

#Plotting the Points
plt.plot(xi,yi,color='green', label='Curve')
plt.scatter(X1,Y1,s=0.5,color='blue', label='Points Under the Curve')
plt.scatter(X2,Y2,s=0.5,color='red', label='Points Above the Curve')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Points Inside the area")
plt.legend()
plt.grid()
plt.show()
```




# 5.   **MONTE CARLO DOUBLE INTEGRAL**





*   **HIT OR MISS**



```python
#Importing Required Libraries
from scipy.integrate import dblquad
import numpy as np
import plotly.graph_objects as go

#Limits of Integration by user Input
a=int(input("Enter lower limit for x, a: "))
b=int(input("Enter Upper limit for x, b: "))
c=int(input("Enter lower limit for y, c: "))
d=int(input("Enter Upper limit for y, d: "))

#Defining function whose Integral is to be found
def f(x,y):
  return 4*x*y-x**2-y**2

nc=0
nt=10000 #Total Number of Points

#To find the maximum value of y
xi=np.linspace(a,b,nt)
yi=np.linspace(c,d,nt)
xi, yi = np.meshgrid(xi, yi)
zi=f(xi, yi)
M=np.max(zi)

X1=[]
X2=[]
Y1=[]
Y2=[]
Z1=[]
Z2=[]

for n in range(1,nt + 1):
  x=np.random.uniform(a,b)
  y=np.random.uniform(c,d)
  z=np.random.uniform(0,M)
  if z<=f(x,y):             #Required Condition for Monte Carlo Simulation
    nc+=1                   #Points Under the Curve
    X1.append(x)
    Y1.append(y)
    Z1.append(z)
  else:
    X2.append(x)
    Y2.append(y)
    Z2.append(z)

func=(b-a)*M*(nc/nt)
print("Integral Solution from monte carlo: ", func)
c,_=dblquad(f, a, b, c, d) #Solving analytically using dblquad function
Error=(abs(func-c)/c)*100  #To calculate the associated Percentage Error
print("Analytical Solution: ", c)
print(f"Error: {Error}%")

#3d Plot of Random Numbers
fig = go.Figure(data=[go.Scatter3d(x=X1, y=Y1, z=Z1,
                                   mode='markers', name='Points Under the Curve', marker=dict(color='Blue', size=2)), go.Scatter3d(x=X2, y=Y2, z=Z2,
                                   mode='markers', name='Points Above the Curve', marker=dict(color='Red', size=2))])
fig.update_layout(
        title='Points Inside the Volume',
        scene = dict(
            xaxis = dict(title='X'),
            yaxis = dict(title='Y'),
            zaxis = dict(title='F(X,Y)')
        )
    )
fig.show()
```

```python
#Importing Required Libraries
from scipy.integrate import dblquad
import numpy as np

#Limits of Integration by user Input
a=int(input("Enter lower limit for x, a: "))
b=int(input("Enter Upper limit for x, b: "))
c=int(input("Enter lower limit for y, c: "))
d=int(input("Enter Upper limit for y, d: "))

#Defining function whose Integral is to be found
def f(x,y):
  return 4*x*y-x**2-y**2

nt=10000 #Total Number of Points
z=0
for n in range(1,nt + 1):
  x=np.random.uniform(a,b)
  y=np.random.uniform(c,d)
  z+=f(x,y)

func=(b-a)*(d-c)*z/(nt)
print("Integral Solution from monte carlo: ", func)
c,_=dblquad(f, a, b, c, d) #Solving analytically using dblquad function
Error=(abs(func-c)/c)*100  #To calculate the associated Percentage Error
print("Analytical Solution: ", c)
print(f"Error: {Error}%")
```

2. Integration from MCMC Sampling

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the target function (up to a constant)
def f(x):
    return np.exp(-x**2) * (2 + np.sin(5 * x) + np.sin(2 * x))

# Metropolis-Hastings MCMC
def metropolis(f, x0, steps, proposal_std):
    samples = []
    x = x0

    for _ in range(steps):
        x_new = np.random.normal(x, proposal_std)
        alpha = f(x_new) / f(x) if f(x) != 0 else 1
        u = np.random.uniform(0, 1)
        if alpha >= 1 or u < alpha:
            x = x_new
        samples.append(x)
    return np.array(samples)

# Parameters
x0 = 0.0               # Starting point
steps = 10000          # Number of MCMC steps
proposal_std = 0.5     # Standard deviation for proposal distribution

# Run MCMC
samples = metropolis(f, x0, steps, proposal_std)

# Plotting the target function
x_vals = np.linspace(-5, 5, 1000)
y_vals = f(x_vals)

plt.figure(figsize=(6, 6))

# Plot the target function
# Histogram of samples
plt.hist(samples, bins=50, density=True, alpha=1, label='Sample Histogram')
plt.plot(x_vals, y_vals / np.trapz(y_vals, x_vals), 'r-', label='Normalized $f(x)$')
plt.title("MCMC Sampling Results")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

# ---------------------------------------------
# Function to integrate
# ---------------------------------------------
def f(x):
    return np.sin(x)

# ---------------------------------------------
# User Inputs
# ---------------------------------------------
a = float(input("Enter the lower limit (a): "))
b = float(input("Enter the upper limit (b): "))
count = int(input("Enter total number of random points: "))
step = int(input("Enter step size (e.g., 100): "))

actual_integral, _ = quad(f, a, b)


x_vals = np.linspace(a, b, 1000)
y_vals = f(x_vals)
y_min = min(0, np.min(y_vals))
y_max = max(0, np.max(y_vals))
rectangle_area = (b - a) * (y_max - y_min)


x_inside, y_inside = [], []
x_outside, y_outside = [], []

estimates, errors, point_counts = [], [], []
accepted_count = 0

for n in range(1, count + 1):
    x_rand = np.random.uniform(a, b)
    y_rand = np.random.uniform(y_min, y_max)
    f_val = f(x_rand)

    if (0 <= y_rand <= f_val) or (0 >= y_rand >= f_val):
        accepted_count += 1
        x_inside.append(x_rand)
        y_inside.append(y_rand)
    else:
        x_outside.append(x_rand)
        y_outside.append(y_rand)


    if n % step == 0:
        estimate = (accepted_count / n) * rectangle_area
        error = abs(estimate - actual_integral)
        estimates.append(estimate)
        errors.append(error)
        point_counts.append(n)

plt.figure(figsize=(8, 6))

plt.scatter(x_inside, y_inside, color='blue', s=5, label='Inside Curve')
plt.scatter(x_outside, y_outside, color='red', s=5, label='Outside Curve')
plt.plot(x_vals, y_vals, color='black', label='f(x) = sin(x)', linewidth=1.5)
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Monte Carlo Samples")
plt.legend()
plt.grid(True)
31
plt.tight_layout()
plt.show()


print(f"\nEstimated Integral after {count} points: {estimates[-1]:.6f}")
print(f"Actual Integral (using quad): {actual_integral:.6f}")
print(f"Final Absolute Error: {errors[-1]:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return 1/(1+x**2)

# Integration range
a, b = -3, 3
true_integral = np.arctan(b) - np.arctan(a)

# Monte Carlo Simulation
N = 400
x_rand = np.random.uniform(a, b, N)
y_rand = f(x_rand)

approx_means = [np.mean(y_rand[:i]) for i in range(1, N+1)]
approx_integrals = [(b-a)*m for m in approx_means]
errors = [abs(I - true_integral) for I in approx_integrals]

# Create side-by-side plots
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# ---- Left plot ----
x = np.linspace(a, b, 500)
axs[0].plot(x, f(x), 'r-', label='f(x) = 1 / (1 + x²)')   # red function line
axs[0].scatter(x_rand, y_rand, s=8, color='blue', label='Sampled Points')  # blue sampled points
axs[0].axhline(np.mean(y_rand), color='green', ls='--',
               label=f'Approximated Mean ≈ {np.mean(y_rand):.4f}')  # green dashed line
axs[0].set_title("Integration by Monte Carlo method")
axs[0].set_xlabel("x")
axs[0].set_ylabel("f(x)")
axs[0].legend()
axs[0].grid()

# ---- Right plot ----
axs[1].plot(range(1, N+1), errors, color='purple')  # error curve in purple
axs[1].set_title("Deviation from True Integral vs Sample Size")
axs[1].set_xlabel("Number of Samples")
axs[1].set_ylabel("Absolute Error")
axs[1].grid()

plt.tight_layout()
plt.show()
```
